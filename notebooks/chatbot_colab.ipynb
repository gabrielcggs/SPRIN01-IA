```python
import os
from google.colab import userdata

os.environ["LLM_PROVIDER"] = "gemini"
os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
os.environ["GEMINI_MODEL"] = "gemini-2.5-flash-lite"
```

```python
SYSTEM_PROMPT = """
Você é o ChargeGrid Intelligence da GoodWe (Contexto A — EV Challenge 2026).
Dados simulados: CG-01 Disponível, CG-02 Em uso 68%, CG-03 Manutenção, CG-04 Disponível.
Consumo junho/2026: 847 kWh (R$ 753,83). Demanda: 38/50 kW sem sobrecarga. Tarifa: R$ 0,89/kWh.
Responda em PT-BR, de forma profissional. Não invente dados fora deste contexto.
"""

from google import genai
from google.genai import types

client = genai.Client(api_key=os.environ["GOOGLE_API_KEY"])
chat = client.chats.create(
    model="gemini-2.5-flash-lite",
    config=types.GenerateContentConfig(system_instruction=SYSTEM_PROMPT),
)

def perguntar(texto):
    r = chat.send_message(texto)
    print(f"Você: {texto}")
    print(f"ChargeGrid: {r.text}\n")
    return r.text
```

```python
# Teste 1 — consumo mensal
perguntar("Quanto gastei este mês em recarga?")
```

```python
# Teste 2 — disponibilidade (memória: continuação da conversa)
perguntar("Qual carregador está disponível agora?")
```

```python
# Teste 3 — custo no CG-02
perguntar("Quanto custa carregar meu veículo no CG-02?")
```

```python
# Teste 4 — status do CG-02
perguntar("Meu carregamento no CG-02 terminou?")
```

```python
# Teste 5 — sobrecarga energética
perguntar("Existe sobrecarga energética agora no condomínio?")
```
